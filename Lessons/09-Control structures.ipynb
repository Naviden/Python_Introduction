# Control Flow

Until now, we've worked with single items. For example, slicing a string or writing to a file. But what if we want to perform an action on a sequence of objects? That's where loops come in. The first type of loop we'll explore is the `for` loop.

## For statement

Suppose that we have a list of our colleagues year of births and we want to calculate their ages, print them and put them in a new `list` called `ages`:

```python
years = [1987, 1990, 1980, 1973]
ages = [] # creating an empty list
```

We know how to get a list's items using indexes. So what if we take each item and subtract it from 2019?

```python
age_1 = 2019 - years[0]
print(age_1)
ages.append(age_1)

age_2 = 2019 - years[1]
print(age_2)
ages.append(age_2)

age_3 = 2019 - years[2]
print(age_3)
ages.append(age_3)

age_4 = 2019 - years[3]
print(age_4)
ages.append(age_4)
```

```python
print(ages)
```

Okay, we did what we were asked to do. But let’s be honest, it wasn’t very efficient! We wrote more than 12 lines of code just to perform three simple operations on four items. Now imagine you need to calculate the age of all Bicocca employees! Luckily for repetitive tasks like this, especially when working with sequence objects like lists, tuples, dictionaries, or strings, we can use loops. In this case, we’ll use a `for` loop:

```python
# using a simple for loop
for i in range(len(years)):
    year = years[i]
    print(2025 - year)
```

We can also use an advanced `for` loop:

```python
ages = []
for year in years:
    age = 2019 - year
    print(age)
    ages.append(age)
```

Notice that the scope of the variables created in the for loop declaration is not limited inside the `for`:

```python
print(i)
print(year)
```

## While statement

`While` loops have a similar purpose to `for`, but there’s a key difference: a `for` loop typically iterates through all the items in a given sequence, while a `while` loop continues running until a specific condition is no longer true.

Suppose we want to print the numbers from 0 to 10 using a `while` loop. Let’s break down what we know from the question:

- we should print numbers
- numbers should be between 0 and 10

```python
num = 0 # initial number is set to 0
while num <= 10: # while the number is less than or equal 10, continue
    print(num, end='-')
    num = num + 1 # adding one to the current value of the num = num+=1
```

Just like any other thing, we can achieve the same result using different methods and functions. For example we can get the numbers we want by using a `for` loop:

```python
for num in range(11): # what is range?
    print(num, end='\t')
```

*Note: Just like for loops, `while` loops also can have a `else` clause*

## If statement

Often, you need to execute some statements only if some condition holds, or choose statements to execute depending on several mutually exclusive conditions. The Python compound statement `if`, which uses `if`, `elif`, and `else` clauses, lets you conditionally execute blocks of statements. Here’s the syntax for the `if` statement:

<code>if_stmt ::= "if" assignment_expression ":" suite
            ("elif" assignment_expression ":" suite)*
            ["else" ":" suite]
</code>

It's common that you want to execute a part of code if a certain condition holds (it's `True`). To do so in Python we use the `if` statement. Let's go back to the simple example of printing numbers we've seen for `while` and `for` statements, but this time suppose that you want to print the number only if it's odd:

```python
num = 0 # initial number is set to 0
while num <= 20: # while the number is less than or equal 20, continue
    if num%2 != 0: # if the reminder of them with 2
        print(num)
    num = num + 1 # adding one to the current value of the num = num+=1
```

Ok, now let's use also `else` and `elif` clauses. Imagine that you have a set of ages and you want to print one if the following labels based on the age:
- kid
- Teen
- adult
- old

```python
ages = [56, 2, 14, 8, 33, 23, 19, 80, 6, 9, 27]

for age in ages:
    if age <= 12:
        print(age, 'Kid', sep=' - ')
    elif age <= 19 and age >= 13:
        print(age, 'Teen', sep=' - ')
    elif age <= 60 and age >= 20:
        print(f'{age} - Adult') # Did you noticed anything new here?
    else:
        print(age, 'Old', sep=' - ')
```

## Break, Continue and pass clauses

Let's have a quick review of statements we have learned until now:
- **`For` statement** : repeats a task for a given sequence
- **`While` statement** : repeats a task while a given condition is `True`
- **`If` statement** : Evaluates a condition and acts upon the result

Looking at these statements it's clear that we can combine them to get a certain task done. For instance: for each item in this list, check if condition 1 is `True` do action 1, otherwise do action 2:

```python
salaries = [1000000, 5000, 11000]

for salary in salaries:
    if salary < 120_000:
        print(f'{salary:,} : Low')
    elif salary > 999_000:
        print(f'{salary:,} : High')
    else:
        print(f'{salary:,} : Normal')
```

In the above example we used `for` and `if` together to examine each item and print a label based on the item's value. Well, we don't always need to iterate through every item of a sequence. Sometimes we need to break the loop, or continue to the next item without spending time on the current item. We can achive these types of action by using these clauses:

- `continue`
- `break`
- `pass`

Suppose you want to find the first number that is less than a specific number in any given list of numbers. You can create a function to do this using a `for` loop:

```python
def find_number(data, limit):
    """Finds the first number in a given list which is less than the defined threshold
    
    Args:
        data : A list containing numerical values
        limit : Threshold defined by user
    
    Returns:
        The first number in the given list smaller than threshold
    """
    all_good = []
    for i in data:
        if i < limit:
            answer = i
            all_good.append(i)
    return all_good
```

Ok, let's test our function:

```python
nums = [973, -1428, 315, -788, 1477, -428, -1187, 592, -108, 370]
limit = -500

print(find_number(nums, limit))
```

Well, it kinda works...the problem is that it returns the last number in the list that is less than -500, instead of the first one! (-1428)
Of course this is not because Python couldn't find the first number (-1428): the problem is that, since we simply used a `for` loop, after finding -1428 Python continued to examine other items in the list... something that we shouldn't have done. There are cases like this that we want to stop (`break`) the loop as soon as a certain condition is satisfied. 

We can write our function like this:

```python
def find_number(data, limit):
    """Finds the first number in a given list which is less than the defined threshold
    
    Args:
        data : A list containing numerical values
        limit : Threshold defined by user
    
    Returns:
        The first number in the given list smaller than threshold
    """
    for i in data:
        if i < limit:
            answer = i
            break
    return answer

print(find_number(nums, limit))
```

And as you can see this time we get the right answer. In this simple case we could have achieve this also without using `break` and just by putting `return` in a different place:

```python
def find_number(data, limit):
    """Finds the first number in a given list which is less than the defined threshold
    
    Args:
        data : A list containing numerical values
        limit : Threshold defined by user
    
    Returns:
        The first number in the given list smaller than threshold
    """
    for i in data:
        if i < limit:
            answer = i
            return answer

print(find_number(nums, limit))
```

The reason that the function above gives us the answer, even without using `break`, is that when Python reaches a `return` statement in a function, it exits immediately without continuing with the rest of the code.
Let's do another example: suppose you received 4 orders for a can of beer from 4 of your customers. In order to know if you can proceed with the order or not, you need to check 3 things: client having at least 18 years old, having at least 2 Euros in his bank card and be in a 3 kilometer radius from your shop. As you can see in the following cell, we can use `continue` after the age check. In this way, if the client has less than 18 years old, it's useless to check for the other conditions:

```python
clients = {'2341': {'age': 15, 'name': 'Ale', 'balance': 658, 'distance': 11},
           '5682': {'age': 61, 'name': 'Ettore', 'balance': 890, 'distance': 1},
           '1873': {'age': 19, 'name': 'Tacca', 'balance': 121, 'distance': 5},
           '9950': {'age': 31, 'name': 'Navid', 'balance': 0, 'distance': 12}}

codes = list(clients.keys())

for code in codes:
    if clients[code]['age'] < 18:
        continue
    if clients[code]['balance'] >= 2 and clients[code]['distance'] < 3:
        print(f"{clients[code]['name']} can have a beer!")
```

Now imagine that the shop manager tells you that he is thinking to do something for the clients that have all the conditions (age and credit) but are living in more distant neighbourhoods. The problem is that it's Fridays evening, you want to go home as soon as possible, but he can't make up his mind about how exactly he's going to handle the delivery situation. So you know you need to modify your function to reflect the new situation, but you still don't know what the solution would be. You can use a `pass` clause as a *placeholder* in your function:

```python
clients = {'2341': {'age': 15, 'name': 'Ale', 'balance': 658, 'distance': 11},
           '5682': {'age': 61, 'name': 'Ettore', 'balance': 890, 'distance': 1},
           '1873': {'age': 19, 'name': 'Tacca', 'balance': 121, 'distance': 5},
           '9950': {'age': 31, 'name': 'Navid', 'balance': 0, 'distance': 12}}

codes = list(clients.keys())

for code in codes:
    if clients[code]['age'] < 18:
        continue
    if clients[code]['balance'] >= 2:
        if  clients[code]['distance'] < 3:
            print(f"{clients[code]['name']} can have a beer!")
        else:
            pass
```
