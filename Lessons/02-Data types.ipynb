## Booleans

Booleans are a core data type of Python, and they consist of only two values: **True** and **False**. Boolean values are mainly used when comparing different values together. (More on comparisons later).
In Python, True has a value of 1 while False's value is 0. We can check this fact by converting them to integer:

```python
print(int(True))
print(int(False))
```

The int function is used to parse (or convert) compatible objects into integers. On the countrary, every non-zero value (or object different from None) is mapped to True:

```python
print(bool(100))
print(bool({"a": 1}))
print(bool(0))
print(bool(None))
```

This is useful for example to check the result of multiple evaluations:

```python
expected_false = False + False + False + False
print(f"The sum of any number of False results should be zero:", expected_false)
expected_true = True + True + True 
print(f"The sum of 3 True results should be 3:", expected_true)
```

### Integer and Float operations

Let's do some basic arithmatic and see the result data types:

```python
a = 1_869 + 67.34
print(a)
print(type(a))
```

```python
a = 1 * 67.34
print(a)
print(type(a))
```

```python
a = 12.5 + 10.5
print(a)
print(type(a))
```

```python
a = 12. / 2
print(a)
print(type(a))
```

To summarize:
- **int** (+ or - or \*) **int** = **int**
- **float** (+ or - or \*) **float** = **float**
- **int** / **int** = **float**
- **int** / **float** = **float**
- **float** / **float** = **float**

## Checking the data types

There are two main way to check the data type of a object. The simplest one is to use `type`, as we see at the beginning of this notebook:

```python
print(type(4524))
print(type('a very good dog'))
```

using this method, we can check if a variable has a certain type:

```python
type([3, [], {'foo': 'bar'}]) == dict
```

A more Pythonic way to do so is to use `.isinstance` built-in function:

```python
isinstance('49.890', str) # checking if '49.890' is a string
```

The main advantage of this function respect to the `type` is that we can use it to do multiple checks:

```python
age_inputs = [11, 14, 13, '12', 12, [16]]

for age in age_inputs:
    if not isinstance(age, (int, str) ):
        print(f'{age} is {type(age)}. It should be either integer or string')
```
