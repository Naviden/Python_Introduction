# Strings

Strings can hold almost anything in character form, including text, symbols, and digits.

As an example, let's take a look at a list (more on lists later) containing 7 strings:

```python
[
'a simple text',
"R3a11ySm4rTP@$swo0rD", 
'navidnobani',
'1366',
'7945-1414 box',
"cane",
"""I really hate these 
        !@$&ing trains"""
]
```

Roughly speaking, we can define strings as anything enclosed in `'`, `"`, or `"""` (though this is a simplified explanation). You might ask, why are there three different ways to define a string? Good question! While they all produce the same type of string object (for example, `'cane'`, `"cane"`, and `"""cane"""` are identical in Python), each format is useful in different situations.

Let’s say we want to create the following string:

> **I can't go to work today**

Let’s first try using double quotes:

```python
"I can't go to work today"
```

This works perfectly! And now `"""`:

```python
"""I can't go to work today"""
```

Another success! And finally, using single quotes:

```python
'I can't go to work today'
```

Wait, what is **invalid syntax** now! We get such an error when Python thinks what you have written doesn't make sense! To better understand what a **Syntax Error** is, let's see a natural language example:

- This is a correct sentence $\rightarrow $ *Marco easts sandwich* $\rightarrow $ (noun) (verb) (noun)
- This is a **Syntactically wrong** sentence $\rightarrow $ *Marco cat sandwich* $\rightarrow $ (noun) (noun) (noun)

So does it mean any (noun) (verb) (noun) gives us a correct combination? Yes and No! Yes because (noun) (verb) (noun) is a **Syntactically correct** sentence but it might still be meaningless $\rightarrow $ Marco listens sandwich (it has a **static semantic error**)

The same thing exists in Python:
- *'Seven'11* is Syntactically wrong because Python can't  understand what do you mean by (str)(int) $\rightarrow $ **SyntaxError**
- *'seven' * 'two'* is also wrong but not Syntactically because although Python can understand what is your goal, (multiplying two strings) it can't do it $\rightarrow $ **TypeError**

Back to our example in the last cell, can you spot the problem? (Of course you can!)
We’re dealing with a string that contains quotes, and Python gets confused about where the string starts and ends. There are two common ways to fix this:

- Use alternative quote styles: for example, use single quotes '...' if the string contains double quotes, or vice versa.
- Use an escape character: In Python (and many other languages) some characters have a reserved role. To tell Python to treat these characters as regular text, we use a backslash (`\`). So if you want to include a quote inside a string, you write it like this: `\'` or `\"`:

```python
'I can\'t go to work today'
```

Now let's try to write a multi-line string :

```python
'some
thing
long'
```

```python
"some
thing
long"
```

```python
print("""some
thing
long""")
```

Therefore, triple quotes can be used for multi-line strings!
There are other ways to insert endlines in strings, for example:

```python
print('just\nwrite\tsomething')
```

Well, it seems to be working but what are those weird `\n` s in the middle? `\n` is one of the [Python literals](https://docs.python.org/3/reference/lexical_analysis.html#escape-sequences) that is used to indicated a new line. We don't go into details here but remember what we have said before about the characters with special meaning and how sometimes we need to escape them. As the last example, let's try to make a string representing a file's absolute path:

> `C:\Documents\Newsletters\Summer2018.pdf`

```python
print('C:\Documents\Newsletters\Summer2018.pdf')
```

As you can see we're getting a SyntaxError again. The reason is that Python thinks we want to use `\n` but we wrote `\N` instead. Since our goal here is not to indicate a new line (`\n`) but to use `\` as part of the string, we need to escape this special character. How can we do it?

By using `\` itself! When Python sees `\\` it understands that you don't want to use special meaning of `\` but you actually need to use backslash itself:

```python
'C:\\Documents\\Newsletters\\Summer2018.pdf'
```

## Common string operations

Having a string, you can apply a wide variety of string methods in order to modify it. There are dozens of operations which you can find [here](https://docs.python.org/3/library/string.html). We'll see together the mostly used ones.

### Split

Returns a list of substrings, cutting the original string at the indicated character:

```python
text = 'This is a Python course'
print(text)
text = text.split(sep=' ')
print(text)
```

The "sep" indication denotes an optional variable for the split function. The default separation will split on any whitespace - use <code>help(str.split)</code> for details. Changing the separation characted:

```python
text = 'I saw the guy, and suddenly he disappeared'
print(text)
text = text.split(',') # notice that I wrote ',' instead of sep=','
print(text)
```

### Slicing

Cutting the string in a way to have new string which is the subset of the original string.

Actually, before explaining slicing, we need to learn a principal concept in Python : Indexing.
Suppose you have a box with 11 pens in it. If you want to take out them one by one and count them, most probably you'll go with:
- One
- Two
- ...
- Eleven

It means, you're using a 1-based index, you indicate the first item with number 1. Python on the other hand, uses a 0-based index. It means it refers to the first object as number(index) 0. So to count the pens in our example, Python goes:
- Zero
- One
- ...
- ten

Ok, now that we know how Python indexing works, we can see string indexing.
In Python Indexing/slicing is not limited to strings, instead, we can slice/index any object that can be considered as a group of objects. That being said, It should be clear now that for example the string: "cane" while considered as a string object is also seen as a container(not as a technical definition) of "c", "a", "n" and "e" with indexes 0, 1, 2 and 3. It means we can slice this string using their indexes. Let's see an example:

suppose we have this string: "NEW YORK" and we want to get the first 3 letters. The indexes are show below:

<img src="../Images/slice.png" width="500">



```python
'New York'
```

```python
t = 'New York'
t[4:]
```

```python
text = 'NEW YORK'
print(text[1]) # E: second letter

print(text[0:3]) # NEW
print(text[40:80]) # YORK

print(text[:3]) # NEW
print(text[4:]) # YORK
```

Indexes can be address not only from the beginning but also from the end:

<img src="../Images/slice_2.png" width="500"> 

```python
text = 'NEW YORK'
print(text[-7]) # E: second letter

print(text[-8:-5]) # NEW
print(text[-4:]) # YORK

print(text[:-5]) # NEW
```

### Replace

Returns a new version of the original string in which the `str_to_be_replaced` instances are replaced with `str_to_be_replaced_with`:

```python
text = 'I love my dog and my little dog'
print(text.replace('dog', 'cat', 1))
```

```python
t = 'tavolo cavolo'
t.replace('cavolo', 'gatto')
```

### Concatenation

Concatenation allows you to create new strings by concatenating existing string together, using `+`:

```python
text_1 = 'Hello'
text_2 = 'World!'
text_3 = text_1 + ' ' + text_2

print(text_3)
```

### Strip

Removes the given string from the beginning and end of the original string. Note: like the `split()` function, the default value of `strip()` is the whitespace, so if you run `my_text.strip()`, it cleans the `my_text` string from starting and ending spaces.

```python
text_1 = ' Pirelli Spa '
print(text_1.strip())

text_2 = 'Continental -'
print(text_2.strip('-').strip())

text_3 = '*Python***'
print(text_3.lstrip('*')) # There is also "rsplit"
```

Although not string operations, `print` and `len` function are Python built-in functions (more on this later) which can be used with different data types like strings, numbers, lists,...

### Printing a string

The `print` function prints(!) the input to the `stdout` (standard output). Note that not all the Python objects can be printed out.

```python
print('something')
print('-'*40)

text = 'just a short text'
print(text)
print('-'*40)

text_1 = 'My name is'
text_2 = 'Navid'
print(text_1, text_2)
print('-'*40)

text_3 = 'University'
text_4 = 'of'
text_5 = 'Bicocca'
print(text_3, text_4, text_5)
print('-'*40)

text_6 = 'folder1'
text_7 = 'folder2'
text_8 = 'folder3'
text_9 = 'file'
print(text_6, text_7, text_8, text_9, sep='/')
```

### Other operations

There are some operations that perform a transformation on strings:

```python
text = 'MilaN'
print(text.lower())
print(text.upper())
print(text.title())
```

Being a strongly typed language, Python doesn't let you to use operations on objects unrelated to them. For example "'Cat' + 3" gives you a `TypeError` since addition operation is for numerical objects not strings:

```python
'cat' - 3
```

But at the same time, there are some exceptions. For example "'cat' * 3" works since Python assume that you want to repeat the string 'cat' , 3 times:

```python
'cat' * 3
```
