```python
import pandas as pd
```

```python
an = pd.read_csv('anagrafica_aziendale.csv', sep='\t')
ges = pd.read_csv('gestionale_paghe.csv')
si = pd.read_csv('sistema_hr.csv', sep=';')
```

```python
si.head()
```

```python
d_an =set(an['INDIRIZZO_EMAIL'].unique())
d_ges =set(ges['email'].unique())
d_si =set(si['Email'].unique())
```

```python
dipendenti = d_an.union(d_ges).union(d_si)
```

```python
len(dipendenti - d_an)
```

```python
len(dipendenti - d_ges)
```

```python
len(dipendenti - d_si)
```

```python
total_dip = []

for email in dipendenti:
    temp_d = {email: {
        'anagrafica': 0,
        'gestionale': 0,
        'sistema_hr': 0
    }}
    if email in d_an:
        temp_d[email]['anagrafica'] = 1
    if email in d_ges:
        temp_d[email]['gestionale'] = 1
    if email in d_si:
        temp_d[email]['sistema_hr'] = 1

    total_dip.append(temp_d)
```

```python
total_dip = []

for email in dipendenti:
    temp_d =[email]
    if email in d_an:
        temp_d.append(1)
    else:
        temp_d.append(0)


    if email in d_ges:
        temp_d.append(1)
    else:
        temp_d.append(0)

        
    if email in d_si:
        temp_d.append(1)
    else:
        temp_d.append(0)

    total_dip.append(temp_d)
```

```python
df = pd.DataFrame(total_dip)
```

```python
df.columns = ['email', 'anagrafica', 'gestionale', 'sistema_hr']
```

```python
df['count'] = df['anagrafica'] + df['gestionale'] + df['sistema_hr']
```
